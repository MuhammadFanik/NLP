# Number Removal

```python
import re
```

```python
text = "I bought 3 apples for $5 in 2024, and the price rose by 15%."
```

```python
# Naive approach: remove all the digit characters using regex
# finds all matches of the pattern and replaces them — here, \d+ matches one or more consecutive digits, and we replace each match with an empty string (''), effectively deleting it
cleaned_text = re.sub(r'\d+', '', text)
print(cleaned_text)
```

```python
# A better alternative - replacing with a placeholder instead of deleting
placeholder_text = re.sub(r'\d+', '<NUM>', text)
print(f"Placeholder version:", placeholder_text)

# This preserves the signal that a number existed (useful for many classification tasks) without keeping every distinct numeric value as a separate, mostly-useless vocabulary token.
```
