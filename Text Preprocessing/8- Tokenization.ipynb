```python
import nltk
nltk.download("punkt")
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
text = "Dr. Smith went to Washington. He arrived at 5 p.m. and said, \"It's freezing!\""
```

## Sentence Tokenization

```python
# sent_tokenize() correctly avoids splitting at "Dr." — it recognizes common abbreviations and doesn't treat that period as a sentence boundary, exactly the theory point about naive period-splitting failing on abbreviations
sentences = sent_tokenize(text)
print(sentences)
```

```python
print("Sentences:")
for s in sentences:
    print(" -", s)
```

## Word Tokenization

```python
# "Washington." became two tokens ['Washington', '.'] (period correctly separated from the word), while "p.m." stayed as one token 'p.m.'
print("\nWord tokens per sentence:")
for s in sentences:
    words = word_tokenize(s)
    print(" -", words)
```
