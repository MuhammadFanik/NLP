# Unicode normalization

```python
import unicodedata
```

```python
# Two visually identical strings, encoded differently
text1 = "café"      # Composed form (single é character)
text2 = "cafe\u0301"    # Decomposed form (e + combining accent)
```

```python
print(text1)
print(text2)
```

```python
print(f"Are they equal? {text1 == text2}")
```

```python
# Normalize both of them to NFC form
normalized1 = unicodedata.normalize("NFC", text1)
normalized2 = unicodedata.normalize("NFC", text2)
```

```python
print(normalized1)
print(normalized2)
```

```python
print(f"Are they equal now after NFC normalization? {normalized1 == normalized2}")
```

## Task

###### Check and print whether word1 == word2 before normalization.
###### Normalize both using NFC, then check and print whether they're equal after normalization.

```python
word1 = "résumé"        # normal composed form
word2 = "re\u0301sume\u0301"        # Decomposed form
```

```python
print(word1)
print(word2)
```

```python
print(f"Are word1 and word2 equal? {word1 == word2}")
```

```python
normalized_word1 = unicodedata.normalize("NFC", word1)
normalized_word2 = unicodedata.normalize("NFC", word2)
```

```python
print(normalized_word1)
print(normalized_word2)
```

```python
print(f"Are they equal now? {normalized_word1 == normalized_word2}")
```
