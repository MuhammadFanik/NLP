```python
import re
```

```python
text = "Check the report at https://finance.com/report?id=123 or email me at analyst@bank.com. Also see @MarketWatch for updates."
```

```python
# Replace URLs with a placeholder
text_no_url = re.sub(r'https?://\S+', '<URL>', text)

# Replace emails with a placeholder
text_no_email = re.sub(r'\S+@\S+\.\S+', '<EMAIL>', text_no_url)

# Replace mentions with a placeholder
text_final = re.sub(r'@\w+', '<MENTION>', text_no_email)

print("Original: ", text)
print("Step 1 (URL):   ", text_no_url)
print("Step 2 (Email): ", text_no_email)
print("Step 3 (Mention):", text_final)
```
