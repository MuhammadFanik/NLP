```python
import nltk
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')  # Only needed once
```

```python
text = "This movie is not good, and I would not recommend it to anyone."
```

```python
words = text.lower().split()
print(words)
```

```python
# stopwords.words('english') gives a predefined list of ~179 common English stopwords. We convert it to a set for fast O(1) membership checking
stop_words = set(stopwords.words('english'))
print(stop_words)
```

```python
# Naive Stopword Removal
filtered_words = [word for word in words if word not in stop_words]
```

```python
print("Original words:", words)
print("Filtered words:", filtered_words)
print("Reconstructed:", " ".join(filtered_words))
```

### A fix - excluding negation words from the stopword list

```python
negation_words = {"not", "no", "never", "n't"}
custom_stopwords = stop_words - negation_words
```

```python
filtered_preserving_negation = [word for word in words if word not in custom_stopwords]
print("Negation-preserving:", filtered_preserving_negation)
```
