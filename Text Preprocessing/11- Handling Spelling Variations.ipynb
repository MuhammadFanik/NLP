```python
import re
```

```python
text = "Sooooo excited!! This is the coloUr of my favourite team's jersey, but it's soooo expensive :("
```

```python
# 1. Collapse elongated characters (3+ repeats -> 2 repeats)
def reduce_elongation(text):
    return re.sub(r'(.)\1{2,}', r'\1\1', text)
```

```python
no_elongation = reduce_elongation(text)
no_elongation
```

```python
# 2. British -> American spelling normalization (simple dictionary mapping)
british_to_american = {
    "colour": "color",
    "favourite": "favorite",
}

def normalize_spelling(text, mapping):
    words = text.split()
    normalized = [mapping.get(word.lower(), word) for word in words]
    return " ".join(normalized)

normalized_text = normalize_spelling(no_elongation.lower(), british_to_american)
```

```python
print("Original:        ", text)
print("Elongation fixed: ", no_elongation)
print("Spelling fixed:   ", normalized_text)
```
