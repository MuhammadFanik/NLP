```python
import re
```

```python
text = "Check out @elonmusk's tweet about #Tesla!! Email me at john@email.com or visit https://tesla.com 🚀"
```

```python
# Remove Emojis
def remove_emoji(text):
    emoji_pattern = re.compile(
        "["
        "\U0001F600-\U0001F64F"  # emoticons
        "\U0001F300-\U0001F5FF"  # symbols & pictographs
        "\U0001F680-\U0001F6FF"  # transport & map symbols
        "\U0001F1E0-\U0001F1FF"  # flags
        "]+", flags=re.UNICODE
    )

    return emoji_pattern.sub('', text)
```

```python
# Remove mentions
no_mentions = re.sub(r'@\w+', '', text)
```

```python
# Strip just the # symbol from hashtags
no_hashtag_symbol = re.sub(r'#(\w+)', r'\1', text)
```

```python
# Remove emojis
no_emoji = remove_emoji(text)
```

```python
print(f"Original: {text}")
print("No mentions:     ", no_mentions)
print("Hashtag stripped:", no_hashtag_symbol)
print("No emoji:        ", no_emoji)
```
