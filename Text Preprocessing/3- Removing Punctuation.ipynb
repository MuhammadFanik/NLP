# Removing Punctuation

```python
import string
```

```python
text = "Wait... what?! Really? Let's go, now!"
```

```python
print(string.punctuation)
```

```python
# Naive approach: remove all the punctuations using str.translate
translator = str.maketrans('', '', string.punctuation)
print(translator)
```

```python
cleaned_text = text.translate(translator)
print(cleaned_text)
```

```python
# string.punctuation is a built-in constant containing all standard ASCII punctuation characters
# str.maketrans('', '', string.punctuation) builds a translation table that maps each punctuation character to None (i.e., delete it)
# .translate(translator) applies that table across the whole string in one efficient pass — much faster than looping character by character or using repeated .replace()
```

```python
print(f"Original Text: {text}")
print(f"Cleaned Text: {cleaned_text}")
print(f"Punctuation characters considered: {string.punctuation}")
```
