```python
import math
```

```python
corpus = [
    "the stock market crashed",
    "the market recovered today",
    "investors are worried about earnings",
    "the earnings report beat expectations"]
```

```python
total_docs = len(corpus)
word = "market"
docs_containing_word = sum(1 for doc in corpus if word in doc.split())
```

```python
idf = math.log(total_docs / docs_containing_word)
```

```python
print(f"IDF for {word} is {idf}")
```
