```python
doc = "the cat sat on the mat"
words = doc.split()
```

```python
words
```

```python
total_words = len(words)
print(total_words)
```

```python
from collections import Counter
word_counts = Counter(words)
```

```python
tf_scores = {word: count / total_words for word, count in word_counts.items()}
```

```python
print(tf_scores)
```
