```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
corpus = [
    "the stock market crashed",
    "the market recovered today",
    "investors are worried about earnings",
    "the earnings report beat expectations"]
```

```python
vc = TfidfVectorizer()
```

```python
tfidf_matrix = vc.fit_transform(corpus)
tfidf_matrix
```

```python
df = pd.DataFrame(tfidf_matrix.toarray(), columns = vc.get_feature_names_out())
print(df.round(3))
```
