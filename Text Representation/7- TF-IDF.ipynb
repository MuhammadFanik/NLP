```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
corpus = [
    "the stock market crashed",
    "the market recovered today",
    "investors are worried about earnings",
    "the earnings report beat expectations"]
```

```python
vc = TfidfVectorizer()
```

```python
tfidf_matrix = vc.fit_transform(corpus)
tfidf_matrix
```

```python
df = pd.DataFrame(tfidf_matrix.toarray(), columns = vc.get_feature_names_out())
print(df.round(3))
```

```python
corpus = [
    "the fed raised interest rates sharply",
    "interest rates are worrying investors today",
    "the central bank cut rates last quarter",
    "investors welcomed the rate cut decision"
]
```

### 1 - Fit a TfidfVectorizer on this corpus, print the resulting matrix as a DataFrame (rounded to 3 decimals).

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
tfidf_vc = TfidfVectorizer()
```

```python
tfidf_matrix = tfidf_vc.fit_transform(corpus)
tfidf_matrix
```

```python
df = pd.DataFrame(tfidf_matrix.toarray(), columns=tfidf_vc.get_feature_names_out())
```

```python
print(df.round(3))
```

### 2 - Identify the word with the highest TF-IDF score in Document 3

```python
# Highest TF-IDF word in Document 3 (index 2)
doc_index = 2
top_word = df.iloc[doc_index].idxmax()
top_score = df.iloc[doc_index].max()
```

```python
print(top_score)
print(top_word)
```
