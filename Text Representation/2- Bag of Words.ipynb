```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
corpus = [
    "the cat sat on the mat",
    "the dog ran in the park",
    "the cat and the dog played"]
```

```python
# Create the vectorizer
vc = CountVectorizer()
```

```python
# Fit on the corpus (learns vocabulary) and transform (produces vectors)
# fit learns the vocabulary from this corpus, transform immediately converts each document into its count vector, using that same vocabulary
bow_matrix = vc.fit_transform(corpus)
bow_matrix
```

```python
# Inspect the vocabulary
# vc.vocabulary_ shows you the learned mapping of word → column index — this is the "dictionary" that gets built during fitting
print(f"Vocabulary: {vc.vocabulary_}")
```

```python
# Convert the sparse matrix into a readable array
# bow_matrix is a sparse matrix by default. .toarray() converts it to a regular dense NumPy array so you can actually see the numbers
print(f"BoW Matrix: {bow_matrix.toarray()}")
```

```python
# See it as a labeled table
import pandas as pd
df = pd.DataFrame(bow_matrix.toarray(), columns=vc.get_feature_names_out())
print(df)
```
