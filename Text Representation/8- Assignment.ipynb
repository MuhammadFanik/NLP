```python
corpus = [
    "the fed raised interest rates sharply",
    "interest rates are worrying investors today",
    "the central bank cut rates last quarter",
    "investors welcomed the rate cut decision",
    "the stock market crashed after the announcement"]
```

# One Hot Encoding

```python
# Step 1: Build the vocabulary — every unique word across the corpus
all_words = []
for sentence in corpus:
    for word in sentence.split():
        all_words.append(word)
```

```python
vocabulary = sorted(set(all_words))
print(vocabulary)
print(f"Vocabulary Size: {len(vocabulary)}")
```

```python
# Step 2: Map each word to its index position in the vocabulary
# enumerate(vocabulary) pairs each word with its position number as it walks through the sorted list. The dictionary comprehension builds a look up table so I can instantly find which position any given word should have its 1 in
word_to_index = {word: idx for idx, word in enumerate(vocabulary)}
```

```python
print(f"Word to Index mapping: {word_to_index}")
```

```python
# Step 3: Build a one-hot vector for a given word.
# This function takes three inputs:
# 1: word --> the specific word you want to encode
# 2: vocabulary --> the full sorted list of unique words
# 3: The look-up dictionary mapping each word to its position
def one_hot_encoder(word, vocabulary, word_to_index):
    vector = [0] * len(vocabulary)  # Start with all zeros. Creates a list of all zeros, exactly as long as te vocabulary is

    # This checks if the given word is in the look-up dictionary
    if word in word_to_index:
        index = word_to_index[word] # looks up that words fixed position
        vector[index] = 1   # flips just that one position to 1, leaving everything else 0

    # Sends back the finished vecotr
    return vector
```

```python
# Try it on 3 words
for word in ["rates", "market", "investors"]:
    vector = one_hot_encoder(word, vocabulary, word_to_index)
    print(f"\nOne-hot vector for '{word}':")
    print(vector)
```

# Bag of Words

```python
corpus = [
    "the fed raised interest rates sharply",
    "interest rates are worrying investors today",
    "the central bank cut rates last quarter",
    "investors welcomed the rate cut decision",
    "the stock market crashed after the announcement"]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
```

```python
# Create the vectorizer
count_vec = CountVectorizer()
```

```python
# Fit on the corpus (learns vocabulary) and transform (produces vectors)
# Fit learns the vocabulary from this corpus, transform immediately converts each document into its count vector, using that same vocabulary
bow_matrix = count_vec.fit_transform(corpus)
```

```python
bow_matrix
```

```python
# Inspect the vocabulary. .vocabulary_ shows you the learned mapping of word → column index — this is the "dictionary" that gets built during fitting
print(f"Vocabulary: {count_vec.vocabulary_}")
```

```python
# Convert the sparse bow_matrix into a readable array. .toarray() converts it to a regular dense NumPy array so you can actually see the numbers
print(f"BoW Matrix: {bow_matrix.toarray()}")
```

```python
# BoW with binary vectors
binary_vec = CountVectorizer(binary=True)
binary_matrix = binary_vec.fit_transform(corpus)
```

```python
df_binary = pd.DataFrame(binary_matrix.toarray(), columns=binary_vec.get_feature_names_out())
print("\nBag-of-Words (binary):\n", df_binary)
```

# N-grams

```python
# Bigrams only
bigram_vec = CountVectorizer(ngram_range=(2, 2))
bigram_matrix = bigram_vec.fit_transform(corpus)
```

```python
df_bigram = pd.DataFrame(bigram_matrix.toarray(), columns=bigram_vec.get_feature_names_out())
```

```python
print(f"Bigram vocabulary: {bigram_vec.get_feature_names_out()}")
```

```python
# Pull out specific meaningful bigrams
for phrase in ["interest rates", "rate cut"]:
    if phrase in df_bigram.columns:
        print(f"{phrase} appears in documents {list(df_bigram.index[df_bigram[phrase] > 0])}")
```
