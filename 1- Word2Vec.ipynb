```python
import gensim.downloader as api
from numpy.ma.core import negative
```

```python
# Load a pretrained model
model = api.load("glove-wiki-gigaword-100")
```

```python
# Basic Vector lookup - first 10 dimensions
# This is the direct indexing like a dictionary returns that word's vector
print(f"Vector for 'king' {model["king"][:10]}")
```

```python
# Most Similar words - Computes Cosine similarity between king and every other word in the vocabulary, and returns the top 10 closest matches.
print(f"Most similar word to king: {model.most_similar("king", topn=10)}")
```

```python
# The famous analogy: king - man + woman
# Positive words get added, negative words get subtracted
result = model.most_similar(positive=["king", "woman"], negative=["man"], topn=5)
print(f"King - Man + Woman = {result}")
```

```python
# Similarity score between 2 specific words
# This returns a single number (cosine similarity ranges from -1 to +1) measuring how close two specific words are.
print(f"Similarity score between king and queen: {model.similarity("king", "queen")}")
print(f"Similarity score between king and apple: {model.similarity("king", "apple")}")
```

```python
# Odd one out
print(f"Odd one out: {model.doesnt_match(["king", "queen", "banana"])}")
```
